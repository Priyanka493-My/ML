```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"
data = pd.read_csv(url)
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
data = data.fillna(0)
```

```python
from sklearn.feature_extraction import DictVectorizer
from sklearn.model_selection import train_test_split
```

```python
df_full_train, df_test = train_test_split(data, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1
)
```

```python
target = 'fuel_efficiency_mpg'
features = ['model_year', 'origin', 'fuel_type', 'drivetrain', 'num_doors',
       'engine_displacement', 'num_cylinders', 'horsepower', 'vehicle_weight',
       'acceleration']


y_train = df_train[target].values
y_val = df_val[target].values
y_test = df_test[target].values

df_train = df_train.drop(columns = [target])
df_val = df_val.drop(columns = [target])
df_test = df_test.drop(columns = [target])
```

```python
dv = DictVectorizer(sparse = False)
train_dict = df_train.to_dict(orient='records')
X_train = dv.fit_transform(train_dict)
```

```python
from sklearn.tree import DecisionTreeRegressor,export_text
```

Question 1

Let's train a decision tree regressor to predict the fuel_efficiency_mpg variable.

* Train a model with max_depth=1.
* Which original feature is used for splitting the data? If you use DictVectorizer, map a one-hot feature such as origin=USA back to its original column name.

```python
dt = DecisionTreeRegressor(max_depth=1)
dt.fit(X_train,y_train)
```

```python
tree_text = export_text(dt,feature_names=list(dv.get_feature_names_out()))
print(tree_text)
```

Question 2

Train a random forest regressor with these parameters:

* n_estimators=10
* random_state=1
* n_jobs=-1 (optional - to make training faster)

What's the RMSE of this model on the validation data?

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error
```

```python
rf = RandomForestRegressor(n_estimators=10,random_state=1,n_jobs=-1 )
rf.fit(X_train,y_train)
```

```python
val_dict = df_val.to_dict(orient='records')
X_val = dv.transform(val_dict)

y_pred = rf.predict(X_val)
rmse = root_mean_squared_error(y_pred,y_val)
print(f'RMSE : {rmse:.3f}')
```

Question 3

Now let's experiment with the n_estimators parameter

* Try these values of this parameter: [10, 50, 100, 150].
* Set random_state to 1.
* Evaluate the model on the validation dataset.

Which value gives the lowest validation RMSE? Consider 3 decimal places for calculating the answer.

```python
n_estimators = [10,50,100,150]
rmses = []
for i in n_estimators :
    rf = RandomForestRegressor(i,random_state=1,n_jobs = -1)
    rf.fit(X_train,y_train)
    y_pred = rf.predict(X_val)
    rmse = root_mean_squared_error(y_pred,y_val)
    rmses.append(rmse)
    print(f'{i} : {rmse:.3f}')

```

Question 4

Let's select the best max_depth:

* Try different values of max_depth: [10, 15, 20, 25]
* For each of these values,
* try different values of n_estimators from [10, 50, 100, 150]
* calculate the mean RMSE
* Fix the random seed: random_state=1

What's the best max_depth, using the mean RMSE?

```python
max_depth = [10, 15, 20, 25]
n_estimators = [10, 50, 100, 150]
rmses = []
for i in max_depth :
    for j in n_estimators:
        rf = RandomForestRegressor(max_depth = i,n_estimators = j,random_state=1,n_jobs=-1)
        rf.fit(X_train,y_train)
        y_pred = rf.predict(X_val)
        rmse = root_mean_squared_error(y_pred,y_val)
        rmses.append(rmse)
        print(f'{i} {j} : {np.mean(rmse):.3f}')        
```

Question 6

Now let's train an XGBoost model! For this question, we'll tune the eta parameter:

* Install XGBoost
* Create DMatrix for train and validation
* Create a watchlist
* Train a model with these parameters for 100 rounds:

```python
import xgboost as xgb
```

```python
features = list(dv.get_feature_names_out())
dtrain = xgb.DMatrix(X_train,label = y_train,feature_names=features)
dval = xgb.DMatrix(X_val,label = y_val,feature_names=features)
```

```python
xgb_params = {
    'eta': 0.3, 
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}
```

```python
watchlist = [(dtrain,'train'),(dval,'val')]
```

```python
model = xgb.train(xgb_params,dtrain,num_boost_round=100,evals=watchlist)
y_pred = model.predict(dval)
root_mean_squared_error(y_val,y_pred)
```

```python
xgb_params = {
    'eta': 0.1, 
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}
```

```python
model1 = xgb.train(xgb_params,dtrain,num_boost_round=100,evals=watchlist)
y_pred1 = model1.predict(dval)
root_mean_squared_error(y_val,y_pred1)
```
